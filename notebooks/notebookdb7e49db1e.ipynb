```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
dane = {
    'Miesiąc': [
        'Styczeń', 'Luty', 'Marzec', 'Kwiecień', 'Maj', 'Czerwiec',
        'Lipiec', 'Sierpień', 'Wrzesień', 'Październik', 'Listopad', 'Grudzień'
    ],
    'Liczba wypadków': [
        127, 103, 130, 144, 166, 149, 
        170, 146, 162, 175, 120, 137
    ]
}
df = pd.DataFrame(dane)

labels = df['Miesiąc'].tolist()
stats = df['Liczba wypadków'].tolist()
num_vars = len(labels)

angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()


stats_closed = stats + [stats[0]]
angles_closed = angles + [angles[0]]

fig, ax = plt.subplots(figsize=(12, 10), subplot_kw=dict(polar=True), facecolor='white')

ax.plot(angles_closed, stats_closed, color='#006d5b', linewidth=2.5, marker='o', markersize=8, label='Liczba wypadków')
ax.fill(angles_closed, stats_closed, color='#006d5b', alpha=0.2)



ax.set_xticks(angles)
ax.set_xticklabels(labels, fontsize=12,fontweight='bold')
ax.set_yticks([50, 100, 150, 180])
ax.set_yticklabels(['50', '100', '150', ''], color='gray', fontsize=9)
ax.tick_params(axis='x', pad=17)
ax.set_rlabel_position(20)
ax.set_ylim(0, 190)
ax.grid(color='lightgray', linestyle='--',alpha=1)

for angle, value, label in zip(angles, stats, labels):
    ax.text(angle, value + 12, str(value), ha='center', va='center', fontsize=13, fontweight='bold', color='#333333')
    
plt.title('Sezonowość śmiertielnych wypadków drogowych (2024)', size=16, y=1.08,fontdict={'fontsize': 16, 'fontweight': 'bold', 'color': '#333333'})
plt.figtext(0.5, 0.95, 
            'Źródło: Raport Biura Ruchu Drogowego Komendy Głównej Policji “Wypadki drogowe w Polsce w 2024 roku”', 
            ha='center', fontsize=12, color='#555555', style='italic')
plt.tight_layout()
plt.savefig('wypadkimiesiace.png', bbox_inches='tight', pad_inches=0.1, dpi=300)
plt.show()
```

```python
dane = {'Rok':[2018,2019,2020,2021,2022,2023,2024],
        'Wypadki':[2747, 2717, 2540, 2488, 2248, 2074,1944]}
df = pd.DataFrame(dane)
df.head()
plt.figure(figsize=(8,6))
plt.title('Wypadki z uczestnikami pod wpływem alkoholu w Polsce (2018-2024)', fontsize=16, fontweight='bold')
ax=sns.barplot(data=dane,x='Rok',y='Wypadki',color='#000080',
        edgecolor='white',
        width=0.6,)
sns.despine(left=True, bottom=False)
ax.yaxis.grid(True, color='lightgray', linestyle='--', alpha=1)
ax.set_axisbelow(True)
for container in ax.containers:
    ax.bar_label(
        container, 
        padding=0.5, 
        fontsize=10, 
        fontweight='bold'
    )
ax.tick_params(axis='x', labelsize=14, pad=10, colors='#444444')
plt.xlabel('Rok', fontsize=12, fontweight='bold')
plt.ylabel('Ilość wypadków', fontsize=12, fontweight='bold')
```
